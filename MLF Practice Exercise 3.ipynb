```python
import pandas as pd
df=pd.read_csv("Customer_orders.csv") 
df
```

```python
# 1. Handle Missing Values

# a. Replace missing CustomerName with "Unknown"
df["CustomerName"] = df["CustomerName"].fillna("Unknown")
df
```

```python
# b. Replace missing Product with "Unknown Product"
df["Product"] = df["Product"].fillna("Unknown Product")
df
```

```python
# c. Drop rows with missing Quantity
df = df.dropna(subset=["Quantity"])
df
```

```python
# d. Replace missing PricePerUnit with the mean
mean_price = df["PricePerUnit"].mean()
df["PricePerUnit"] = df["PricePerUnit"].fillna(mean_price)
df 
```

```python
# 2. Correct Invalid Values

# Convert negative Quantity values into positive values
df["Quantity"] = df["Quantity"].abs()
df
```

```python
# 3. Remove Duplicates

# Identify the number of duplicate rows
duplicate_count = df.duplicated().sum()
df
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
df
```

```python
# 4. Converting Data Types

# Convert Quantity to integer
df["Quantity"] = df["Quantity"].astype(int)
df
```

```python
# 5. Create Derived Column

# Calculate TotalPrice
df["TotalPrice"] = df["Quantity"] * df["PricePerUnit"]
df
```
